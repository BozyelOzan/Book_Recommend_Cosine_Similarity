```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import random
```

```python
df = pd.read_csv("/kaggle/input/goodreads-books-100k/GoodReads_100k_books.csv")
df_copy = df.copy()

df = df[['title', 'author', 'genre', 'rating']]
df.dropna(inplace=True)
df['genre'] = df['genre'].apply(lambda x: x.replace('|', ' '))
```

```python
def combine_features(row):
    return row['title'] + ' ' + row['author'] + ' ' + row['genre']
df['combined'] = df.apply(combine_features, axis=1)
```

```python
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['combined'])
```

```python
def recommend_books(title, top_n=5):
    if title not in df['title'].values:
        return None

    idx = df[df['title'] == title].index[0]
    book_vector = tfidf_matrix[idx]  
    sim_scores = cosine_similarity(book_vector, tfidf_matrix).flatten()
    similar_indices = sim_scores.argsort()[::-1][1:top_n+1]
    return df.iloc[similar_indices][['title', 'author']]
```

```python
def testing_rec(numbers):
    for i in numbers:
        title = df_copy.loc[i, 'title']
        res_recom = recommend_books(title, top_n=6)
        
        if res_recom is None or res_recom.empty:
            print("No similar book found!!!\n")
        else:
            print(f'{title} ----> Here are our recommended books!!!\n')
            print(res_recom)
```

```python
testing_number = [random.randint(1, 100000) for _ in range(5)]

testing_rec(testing_number)
```
